```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\MSII\OneDrive\Desktop\employee-attrition-prediction\data\dataset.csv")
```

```python
df_clean = df.copy()
```

check extra spaces and blank spaces

```python
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()
```

```python
df_clean = df_clean.replace(r"^\s*$", np.nan, regex=True)
```

check missing   values

```python
df_clean.isnull().sum()
```

```python
total_missing = df_clean.isnull().sum().sum()

print("Total missing values:", total_missing)
```

check duplicate records

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
if duplicate_count > 0:
    df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("Shape after duplicate check:", df_clean.shape)
```

```python
unique_counts = df_clean.nunique(dropna=False)

print(unique_counts.sort_values())
```

To Identify Constant columns

```python
constant_columns = unique_counts[unique_counts <= 1].index.tolist()

print("Constant columns:", constant_columns)
```

remove constant columns

```python
df_clean = df_clean.drop(columns=constant_columns)

print("Shape after removing constant columns:", df_clean.shape)
```

Remove Unnecessary Identifier Columns

```python
# first check
if "EmployeeNumber" in df_clean.columns:
    print(
        "Unique Employee Numbers:",
        df_clean["EmployeeNumber"].nunique()
    )
```

```python
# remove colomn
if "EmployeeNumber" in df_clean.columns:
    df_clean = df_clean.drop(columns=["EmployeeNumber"])

print("Shape after removing identifier:", df_clean.shape)
```

to validate Target Column Validate

```python
# check target values
print(df_clean["Attrition"].value_counts(dropna=False))
```

```python
valid_labels = {"Yes", "No"}

actual_labels = set(df_clean["Attrition"].dropna().unique())

unexpected_labels = actual_labels - valid_labels

print("Unexpected target labels:", unexpected_labels)
```

```python
# check misssing value
print(
    "Missing target values:",
    df_clean["Attrition"].isnull().sum()
)
```

final check

```python
df_clean.info()
```

```python
# check missing valiue
print(
    "Total missing values:",
    df_clean.isnull().sum().sum()
)
```

```python
# Final duplicate check
print(
    "Duplicate rows:",
    df_clean.duplicated().sum()
)
```

```python
# Final dataset shape
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
```

```python
from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_path = processed_dir / "employee_attrition_clean.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", output_path)
```
